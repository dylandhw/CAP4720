```python
import pygame
import random
```

```python
pygame.init()

clock = pygame.time.Clock() # create a clock object

display = pygame.display.set_mode((800, 600))
pygame.display.set_caption(title = "Assignment 1")
running = True
```

```python
while running:
    for event in pygame.event.get(): # poll for events
        if event.type == pygame.QUIT or (event.type == pygame.KEYDOWN and event.key == 27):
            running = False
        display.fill((random.randint(0,255), random.randint(0,255), random.randint(0,255)))
        
        pygame.display.flip()
        clock.tick(30) # limits to 30FPS

pygame.quit()
```
