```python
import pygame
import moderngl
import numpy
```

```python
pygame.init()
```

```python
width = height = 600

pygame.display.set_mode((width, height), flags = pygame.OPENGL | pygame.DOUBLEBUF)
pygame.display.set_caption(title = "test window")
gl = moderngl.get_context()
```

```python
gl.info["GL_VERSION"]
```

```python
position_data = [
    0.0, 0.8,
    -0.8, -0.8,
    0.8, -0.8
]
triangle_vertex_positions = numpy.array(position_data).astype("float32")
```

```python
triangle_positions_buffer = gl.buffer(triangle_vertex_positions)
```

```python


vertex_shader_code = '''
#version 330 core

layout (location = 0) in vec2 position; 

void main(){
    gl_Position = vec4(position, 0.0, 1.0);
}
'''
```

```python
fragment_shader_code = '''
#version 330 core

out vec4 outColor;

void main(){
    outColor = vec4(1.0, 1.0, 1.0, 1.0);
}
'''
```

```python
program = gl.program(
    vertex_shader = vertex_shader_code,
    fragment_shader = fragment_shader_code,
)
```

```python
renderable = gl.vertex_array(program, [(triangle_positions_buffer, '2f', 'position')])
```

```python
running = True 

while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT or (event.type == pygame.KEYDOWN and event.key == 27):
            running = False
    gl.clear(0.2, 0.2, 0.2)
    renderable.render()
pygame.display.flip()
```
